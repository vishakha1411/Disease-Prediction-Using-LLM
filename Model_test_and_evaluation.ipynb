```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel

model_name = "mistralai/Mistral-7B-v0.1"   # <-- base model you used

tokenizer = AutoTokenizer.from_pretrained("/content/drive/MyDrive/mistral_lora_adapter")

base = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    load_in_4bit=True,  # same quantization used during training
    trust_remote_code=True
)

model = PeftModel.from_pretrained(
    base,
    "/content/drive/MyDrive/mistral_lora_adapter"
)
```

# **TESTING ON TEST.JSONL AND CONFUSION MATRIX**

```python
from datasets import load_dataset
import torch
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import numpy as np
import re

# Load test dataset
test_ds = load_dataset("json", data_files="/content/test.jsonl")["train"]

# -----------------------------
# Helper functions
# -----------------------------
def is_match(true_label, pred_label):
    t = true_label.lower().strip()
    p = pred_label.lower().strip()

    # exact match
    if t == p:
        return True

    # substring match (handles cases like: "paralysis" vs "paralysis (brain hemorrhage)")
    if t in p or p in t:
        print("Substring match:", t, p)
        return True

    return False

def extract_true_disease(output_text):
    return output_text.split("Disease:")[1].split("\n")[0].strip().lower()

def extract_predicted_disease(generated_text):
    text = generated_text.lower()

    # 1. Try exact "disease: <name>"
    match = re.search(r"disease\s*:\s*([a-z0-9 \-\_]+)", text)
    if match:
        disease = match.group(1).strip()
        # cut at explanation/note/symptoms etc AFTER capturing disease
        for stop in ["explanation", "note", "symptoms", "answer"]:
            disease = disease.split(stop)[0]
        return disease.strip()

    # 2. Try "answer: <name>"
    match = re.search(r"answer\s*:\s*([a-z0-9 \-\_]+)", text)
    if match:
        disease = match.group(1).strip()
        for stop in ["explanation", "note", "symptoms", "answer"]:
            disease = disease.split(stop)[0]
        return disease.strip()

    # 3. Try last line fallback: often disease is the last short line
    lines = [l.strip() for l in text.split("\n") if l.strip()]
    if len(lines) > 0:
        candidate = min(lines, key=len)
        if len(candidate.split()) <= 3:
            return candidate

    return text.strip()


correct = 0
total = 0

true_labels = []
pred_labels = []

for sample in test_ds:
    prompt = (
        f"{sample['instruction']}\n"
        f"Symptoms: {sample['input']}\n"
        f"Answer:"
    )

    true_disease = extract_true_disease(sample["output"])
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=50,
        do_sample=False,
        temperature=0,
    )

    decoded = tokenizer.decode(output[0], skip_special_tokens=True)
    predicted_disease = extract_predicted_disease(decoded)

    true_labels.append(true_disease)
    pred_labels.append(predicted_disease)

    if is_match(true_disease, predicted_disease):
        correct += 1
    else:
        print("\n❌ Wrong Prediction")
        print("Symptoms:", sample["input"])
        print("Expected:", true_disease)
        print("Predicted:", predicted_disease)

    total += 1

accuracy = correct / total
print(f"\n✔ FINAL ACCURACY: {accuracy:.4f} ({correct}/{total})")


labels = sorted(list(set(true_labels + pred_labels)))

cm = confusion_matrix(true_labels, pred_labels, labels=labels)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)

plt.figure(figsize=(10, 8))
disp.plot(xticks_rotation=45, cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# ------------------------------
# true_labels  → list of expected diseases
# pred_labels  → list of model predictions
# Make sure these two lists are already populated earlier
# ------------------------------

# Convert to pandas Series (clean & safe)
true_labels = pd.Series(true_labels)
pred_labels = pd.Series(pred_labels)

# Generate confusion matrix
# Sort labels alphabetically for clean layout
unique_labels = sorted(true_labels.unique())

# Build confusion matrix
cm = confusion_matrix(true_labels, pred_labels, labels=unique_labels)

# Create a DataFrame for better control
cm_df = pd.DataFrame(cm, index=unique_labels, columns=unique_labels)

# ---- Plotting ----
plt.figure(figsize=(14, 10))
sns.heatmap(cm_df, annot=True, fmt="d", cmap="Blues",
            cbar=False, square=True)

plt.xlabel("Predicted", fontsize=14)
plt.ylabel("Actual", fontsize=14)
plt.title("Confusion Matrix — Disease Classification", fontsize=16)
plt.xticks(rotation=45, ha="right", fontsize=11)
plt.yticks(rotation=0, fontsize=11)

plt.tight_layout()
plt.show()

accuracy = (correct / total)*100
print(f"\n FINAL ACCURACY: {accuracy:.2f} % ({correct}/{total})")
```

# **DEMO QUERY**

```python
# demo Query symptoms
symptoms = "Fever, headache, body pain, joint pain"

# Build prompt exactly like training
prompt = (
    "Identify the disease pattern based on symptoms.\n"
    f"Symptoms: {symptoms}\n"
    "Answer:"
)

# Tokenize
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=False,      # deterministic output
    temperature=0,
)

# Decode
decoded = tokenizer.decode(output[0], skip_special_tokens=True)

print("-------- RAW MODEL OUTPUT --------")
print(decoded)

# ----------------------------
# Extract ONLY the final answer
# ----------------------------
def extract_predicted_output(text):
    text = text.lower()
    if "disease:" in text:
        return text.split("disease:")[1].strip()
    return text

clean_output = extract_predicted_output(decoded)

print("\n-------- CLEANED PREDICTION --------")
print(clean_output)
```

# **3 SAMPLE CASES**

```python
test_symptoms = [
    "Fever, headache, body pain, joint pain, nausea",
    "Chest pain, shortness of breath, sweating, fatigue",
    "Vomiting, abdominal pain, yellowish skin, history of alcohol consumption",
    "High fever, chills, nausea, abdominal cramps, diarrhea"
]

def predict(symptoms):
    prompt = (
        "Identify the disease pattern based on symptoms.\n"
        f"Symptoms: {symptoms}\n"
        "Answer:"
    )

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=80,
        do_sample=False,
        temperature=0,
    )

    decoded = tokenizer.decode(output[0], skip_special_tokens=True)

    # Extract only disease
    if "Answer:" in decoded:

        disease = decoded.split("Answer:")[1].strip()
    else:
        disease = decoded

    print("\nPredicted Disease:", disease)
    print("-------------------------------------")

for s in test_symptoms:
    print("\n===============================")
    print("Symptoms:", s)
    predict(s)
```
